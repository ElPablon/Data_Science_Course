```python
import statistics as sts
import pandas as pd
import seaborn as srn
```

```python
dataframe = pd.read_csv('tempo.csv', sep=";")

dataframe.head()
```

```python
#Aparência
dataframe.groupby(['Aparencia']).size()
```

```python
#Remover valore errados de aparência
dataframe.loc[dataframe['Aparencia'] == 'menos', 'Aparencia'] = 'sol'

dataframe.groupby(['Aparencia']).size()
```

```python
#Temperatura
dataframe['Temperatura'].describe()
dataframe.loc[(dataframe['Temperatura'] > 130) | (dataframe['Temperatura'] < -130) ]
```

```python
#Substituir valores fora do dominio pela mediana
medianaTemperatura = sts.median(dataframe['Temperatura'])
dataframe.loc[(dataframe['Temperatura'] > 130) |(dataframe['Temperatura'] < -130), 'Temperatura'] = medianaTemperatura
dataframe.loc[(dataframe['Temperatura'] > 130) |(dataframe['Temperatura'] < -130) ]
```

```python
#Umidade
dataframe['Umidade'].describe()
```

```python
#Substituir valores fora do dominio pela mediana
medianaUmidade = sts.median(dataframe['Umidade'])
dataframe.loc[(dataframe['Umidade'] > 100) | (dataframe['Umidade'] < 0), 'Umidade'] = medianaUmidade
dataframe['Umidade'].describe()
```
